```python
df.columns=(df.columns.str.strip().str.lower().str.replace(" ","_"))
```

```python
text_columns = ["customer_id","customer_unique_id","customer_city","customer_state"]
```

```python
for col in text_columns:
  df[col]=df[col].astype("string").str.strip()
```

```python
df["customer_city"]=(df["customer_city"].str.lower().str.title())
```

```python
df["customer_state"]=(df["customer_state"].str.upper().str.strip())
```

```python
print("Column names after standardization:")
print(df.columns.tolist())
```

```python
print("/n Sample cleaned text:")
display(df[["customer_id","customer_unique_id","customer_city","customer_state"]].head())
```

```python
customer_id_duplicates = df["customer_id"].duplicated().sum()
print(f"Duplicated_customer_id values:{customer_id_duplicates:,}")
```

```python
if customer_id_duplicates == 0:
    print("PASS: customer_id is unique.")
else:
    print("WARNING: customer_id contains duplicates.")
    print("\nDuplicate customer_id records:")
    display(
        df[df["customer_id"].duplicated(keep=False)]
        .sort_values("customer_id")
    )
```

```python
customer_frequency = (df.groupby("customer_unique_id").size().reset_index(name="record_count"))
repeat_customers = customer_frequency[customer_frequency["record_count"]>1]
unique_customers = customer_frequency.shape[0]
print(f"/nUnique_customer_id values:{unique_customers:,}")
print(f"Repeat customer_unique_id values:{unique_customers:,}")
print("/nTop repeat customer IDs:")
display(repeat_customers.sort_values("record_count",ascending=False).head(10))

```

```python
missing_values = df.isna().sum()
duplicate_rows = df.duplicated().sum()

print("Missing values by column:")
print(missing_values)

print(f"\nDuplicate complete rows: {duplicate_rows:,}")

if missing_values.sum() == 0:
    print("PASS: No missing values.")
else:
    print("WARNING: Missing values detected.")

if duplicate_rows == 0:
    print("PASS: No duplicate complete rows.")
else:
    print("WARNING: Duplicate complete rows detected.")
```

```python
df["customer_zip_code_prefix"] = pd.to_numeric(df["customer_zip_code_prefix"],errors="coerce").astype("Int64")
df["customer_zip_code_prefix"] = (
    df["customer_zip_code_prefix"]
    .astype("string")
    .str.zfill(5)
)
print("ZIP code prefix sample:")
display(
    df["customer_zip_code_prefix"].head(10)
)

```

```python
df["customer_id"] = df["customer_id"].astype("string")
df["customer_unique_id"] = df["customer_unique_id"].astype("string")
df["customer_city"] = df["customer_city"].astype("string")
df["customer_state"] = df["customer_state"].astype("string")
df["customer_zip_code_prefix"] = (df["customer_zip_code_prefix"].astype("string"))
print("Final data types:")
print(df.dtypes)

```

```python
assert len(df) == 99_441, (f"Unexpected row count: {len(df):,}")
assert len(df.columns)==5, (f"Unexpected column count: {len(df.columns)}")
expected_columns = [
    "customer_id",
    "customer_unique_id",
    "customer_zip_code_prefix",
    "customer_city",
    "customer_state"]
assert df.columns.tolist()==expected_columns,("Column names/order do not match expected structure.")
assert df["customer_id"].is_unique,("Customer id contains duplicate values")
assert df.isna().sum().sum()==0,("Missing values detected")
assert not df.duplicated().any(),("Duplicate complete rows detected.")
assert df["customer_zip_code_prefix"].str.fullmatch(r"\d{5}").all(), ("ZIP code prefix formatting error detected.")
assert df["customer_state"].str.fullmatch(r"[A-Z]{2}").all(), ("Invalid customer_state value detected.")
print("==========================================")
print("ALL SANITY CHECKS PASSED")
print("==========================================")




```
